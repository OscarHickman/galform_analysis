# Snapshot aggregation

Completed subvolumes per snapshot and total aggregated galaxies with `z`.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / 'galform_analysis').is_dir():
            return candidate
    raise RuntimeError('Repository root with galform_analysis/ not found')

repo_root = find_repo_root(Path.cwd())
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

import polars as pl
import numpy as np

from galform_analysis.analysis.aggregation import completed_galaxies, aggregate_snapshot
from galform_analysis.config import get_base_dir

# Parameters
TARGET_IZ = 'iz155'

# Data location
base_dir = Path(get_base_dir())

# Completion summary (concise)
df_comp = completed_galaxies(basedir=str(base_dir))
summary = df_comp.group_by('iz').agg(pl.len()).sort('iz')
print(summary)

# Aggregate one snapshot and print diagnostics
agg_path = base_dir / TARGET_IZ
agg = aggregate_snapshot(str(agg_path))
n = len(agg['mstar'])
z = agg.get('z', np.nan)
print(f'AGGREGATION: {n:,} galaxies, z={z:.3f}')
```
