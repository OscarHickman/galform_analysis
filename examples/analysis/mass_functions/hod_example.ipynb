```python
from pathlib import Path

def find_repo_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / 'src').is_dir():
            return candidate
    raise RuntimeError('Repository root with src/ not found')

import matplotlib.pyplot as plt
import matplotlib as mpl
from src.analysis.mass_functions.hod import avg_hod_given_redshift_and_subvolumes
from src.config import get_base_dir
import numpy as np

mpl.setconfig()

iz_num, ivols = 155, [0, 1, 2, 3]
res = avg_hod_given_redshift_and_subvolumes(iz_num, ivols, base_dir=str(get_base_dir()))

plt.figure(figsize=(10, 8))

plt.plot(res['centers'], res['mean_occupation'], label='Total')
plt.plot(res['centers'], res['mean_central'], label='Centrals')
plt.plot(res['centers'], res['mean_satellite'], label='Satellites')

plt.yscale('log')
plt.xlabel('log10(M_halo)')
plt.ylabel('<N_gal>')
plt.legend()
plt.show()
```
